# Lexical Similarity Tests

Evaluates TF-IDF lexical similarity against ground-truth scores from the database. Preprocesses articles with lemmatization, computes pairwise TF-IDF cosine similarity, and reports error statistics relative to the human-annotated similarity labels.

```python
from clustering.Cluster import *
from data_access.DataProvider import *
import numpy as np

data_provider = DataProvider()
max_count = 1000
rows = data_provider.get_articles('HR', max_count)

corpus = [row[0] for row in rows[0:max_count]] + [row[1] for row in rows[0:max_count]]
results = [row[2] for row in rows[0:max_count]]
```

```python
from models.Model import *
from textpreprocessing.TextPreprocessing import *
preprocessing = CroatianTextPreprocessor()
tfidf_model = TfIdfModel()
```

```python
results2 = []

for i in range(len(corpus)):
 
    corpus[i] = preprocessing.preprocess_text(corpus[i], apply_lemmatization= True)
    corpus[i] = ''.join(corpus[i])
```

```python
tfidf_matrix = tfidf_model.Transform(corpus, 2, 4)
```

```python
results_t = []

for i in range(max_count):
    results_t.append(tfidf_matrix[i][i + max_count])
    
```

```python
import numpy as np

result_list = [abs(x - y) for x, y in zip(results_t, results)]
```

```python
from utils.utils import *
calculate_stats(result_list)
```
