```python
from sentence_transformers import SentenceTransformer, util

def calculate_bert_similarity(text1, text2):
    # Load a pre-trained BERT model
    model = SentenceTransformer('paraphrase-MiniLM-L6-v2')

    # Encode the texts into BERT embeddings
    embeddings = model.encode([text1, text2], convert_to_tensor=True)

    # Calculate cosine similarity between the embeddings
    similarity_score = util.pytorch_cos_sim(embeddings[0], embeddings[1])

    return similarity_score.item()

# Example texts
text_a = "This is the first document."
text_b = "This document is the second document."

# Calculate BERT similarity
bert_similarity = calculate_bert_similarity(text_a, text_b)

print(f"BERT Similarity between the two texts: {bert_similarity:.4f}")
```
