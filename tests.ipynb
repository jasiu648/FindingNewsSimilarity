```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances_argmin_min

# Sample text data
documents = [
    "Machine learning is fascinating and powerful.",
    "Python is a popular programming language for machine learning.",
    "Clustering helps in organizing similar data points.",
    "K-Means is a common clustering algorithm.",
    "Text similarity is an important concept in natural language processing.",
    "The Python scikit-learn library provides useful tools for machine learning tasks."
]

# Convert the text data into TF-IDF features
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

# Specify the number of clusters
num_clusters = 2

# Apply K-Means clustering
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
kmeans.fit(X)

# Get cluster centroids
centroids = kmeans.cluster_centers_

# Get the index of the closest document to each centroid
closest, _ = pairwise_distances_argmin_min(centroids, X)

# Print the clusters
for i in range(num_clusters):
    cluster_documents = [documents[j] for j in range(len(documents)) if kmeans.labels_[j] == i]
    print(f"Cluster {i + 1}:")
    print(cluster_documents)
    print("\n")

print(X)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import pairwise_distances_argmin_min
import numpy as np

# Sample text data
documents = [
    "Machine learning is fascinating and powerful.",
    "Python is a popular programming language for machine learning.",
    "Clustering helps in organizing similar data points.",
    "K-Means is a common clustering algorithm.",
    "Text similarity is an important concept in natural language processing.",
    "The Python scikit-learn library provides useful tools for machine learning tasks."
]

# Convert the text data into TF-IDF features
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)

# Specify the number of clusters
num_clusters = 2

# Apply Agglomerative Clustering
agglomerative = AgglomerativeClustering(n_clusters=num_clusters)
agglomerative.fit(X.toarray())

# Get the index of the closest document to each cluster center
centroids = np.array([X.toarray()[agglomerative.labels_ == i].mean(axis=0) for i in range(num_clusters)])
closest, _ = pairwise_distances_argmin_min(centroids, X)

# Print the clusters
for i in range(num_clusters):
    cluster_documents = [documents[j] for j in range(len(documents)) if agglomerative.labels_[j] == i]
    print(f"Cluster {i + 1}:")
    print(cluster_documents)
    print("\n")
```
