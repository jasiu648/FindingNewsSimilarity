```python
query = sql.SQL("""
    SELECT
        tr.body,
        ta.body,
        ml.similarity
    FROM
        matching_list ml
    JOIN releases re ON ml.id_releases = re.id 
    JOIN articles ar ON ml.id_articles = ar.id
    JOIN texts ta ON ta.id = ar.id_texts
    JOIN texts tr ON tr.id = re.id_texts
    WHERE ar.language = {};
""").format(sql.Literal(language))

#print(query)
cursor.execute(query)
rows = cursor.fetchall()
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
from collections import Counter
nltk.download('stopwords')

def to_bag_of_words(text, lang):
   
    # Tokenization: Split the text into words or tokens
    tokens = word_tokenize(text)

    # Lowercasing: Convert tokens to lowercase
    tokens = [word.lower() for word in tokens]

    # Removing Punctuation: Remove non-alphanumeric characters
    tokens = [word for word in tokens if word.isalnum()]

    # Removing Stopwords: Remove common words (e.g., 'the', 'is', 'a') that may not carry meaningful information
    stop_words = set([
    "i", "mnie", "moje", "ja", "my", "nasze", "siebie", "ty", "twój", "twój", "tylko", "twój", "twoja", "twoje",
    "siebie", "on", "jego", "jego", "on", "on jest", "jej", "jej", "ona", "ona", "ono", "to", "to jest", "oni", "nich",
    "ich", "oni", "oni", "kto", "który", "który", "komu", "ten", "to", "to będzie", "te", "tamte", "ci", "te", "oni",
    "te"])
    filtered_tokens = [word for word in tokens if word not in stop_words]

    # Stemming: Reduce words to their base or root form
    stemmer = PorterStemmer()
    stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

    # Lemmatization: Reduce words to their base form using WordNet lemmatizer
    lemmatizer = WordNetLemmatizer()
    lemmatized_tokens = [lemmatizer.lemmatize(word, wordnet.VERB) for word in stemmed_tokens]

    word_counts = Counter(lemmatized_tokens)

    return word_counts
```
