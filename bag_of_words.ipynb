```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter

# Sample text data
text = "This is a simple example of a bag of words. The bag of words model represents text as a collection of words."

# Step 1: Tokenization
# Tokenize the text into words
words = word_tokenize(text)

# Step 2: Preprocessing
# Remove stopwords and punctuation, and convert to lowercase
stop_words = set(stopwords.words('english'))
filtered_words = [word.lower() for word in words if word.isalnum() and word.lower() not in stop_words]

# Step 3: Frequency Count
# Create a dictionary of word frequencies
word_counts = Counter(filtered_words)

# Step 4: Display the Bag of Words
# Print the word frequencies
print("Bag of Words (BoW):")
for word, count in word_counts.items():
    print(f"{word}: {count}")

print(word_counts)
```
