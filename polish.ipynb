```python
from models.model import *
from models.ner_models import *
from data_access.DataProvider import *
from textpreprocessing.TextPreprocessing import *

provider = DataProvider()
preprocesor = SlovenianTextPreprocessor()
embedding_model = EnglishBertModel()
rows = provider.get_articles('SL')
ner = SlovenianNER()

corpus = [row[0] for row in rows[0:100]] + [row[1] for row in rows[0:100]]
results = [row[2] for row in rows[0:100]]
```

```python
corpus[0]
```

```python
bert_res = []
for i in range(100):
    bert_res.append(embedding_model.Transform(corpus[i],corpus[i + 100]))
```

```python
bert_res[0]
```

```python
for i in range(len(corpus)):
    corpus[i] = preprocesor.preprocess_text(corpus[i], apply_lemmatization= True)
    corpus[i] = ''.join(corpus[i])
```

```python
corpus[0]
```

```python
tf = TfIdfModel()

tfidf_matrix = tf.Transform(corpus)

tfidf_result = []
for i in range(100):
    tfidf_result.append(tfidf_matrix[0 + i][100 + i])
```

```python
tfidf_result
```

```python
for i in range(len(bert_res)):
    bert_res[i] -= 0.5996
    bert_res[i] *= 100/40
```

```python
from utils.utils import *
calculate_stats(bert_res)
```

```python

ner_results = []
a = []
for i in range(len(corpus)):
    ner_results.append(ner.FindNamedEntites(corpus[i]))
    a.append(([inner_list[0] for inner_list in ner_results[i]]))
```

```python
a
```

```python
jaccard_similarity = []

for i in range(100):
    set1 = set(a[i])
    set2 = set(a[i + 100])
    intersection = len(set1.intersection(set2))
    union = len(set1.union(set2))
    jaccard_similarity.append(intersection / union)

print("Jaccard Similarity:", jaccard_similarity)
```

```python
import openpyxl

# Create a new Excel workbook
workbook = openpyxl.Workbook()

# Select the active sheet (default is the first sheet)
sheet = workbook.active

# Write headers (optional)
headers = ['text1', 'text2', 'lexical','semantic','entities1', 'entities2', 'entitysimilarity']
sheet.append(headers)
data = []

for i in range(100):
    data.append([corpus[i], corpus[i + 100], tfidf_result[i], bert_res[i], ' '.join(a[i]), ' '.join(a[i + 100]), jaccard_similarity[i]])


for row in data:
    sheet.append(row)

# Save the workbook to a file
workbook.save('results_SL.xlsx')
```
